```python
import numpy as np
import matplotlib.pyplot as plt
```

$$ z = Xw + b $$

Donde $ X \in \mathbb{R}^{n\times d} $, $ w \in \mathbb{R}^d, b \in \mathbb{R} $

```python
sigmoid = lambda z: 1 / (1 + np.exp(-z))

X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])
y = np.array([0,0,0,1])
w = np.array([0,0])
b = 0
alpha = 0.2 # learning rate

z= X @ w + b
p = sigmoid(z)

# Binary Cross Entropy (BCE)
def binary_cross_entropy(y_true, y_pred):
    eps = 1e-15  # Para evitar log(0)
    y_pred = np.clip(y_pred, eps, 1 - eps)  # Evitar valores extremos
    loss = -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
    return loss
```

```python
# Número de iteraciones (por ejemplo, 1000)
for i in range(100):
    # Calcular z = Xw + b
    z = X @ w + b
    
    # Calcular probabilidades p (sigmoide de z)
    p = sigmoid(z)
    
    # Calcular la pérdida
    loss = binary_cross_entropy(y, p)
    
    # Calcular gradientes
    dw = (1/m) * X.T @ (p - y)
    db = (1/m) * np.sum(p - y)
    
    # Actualizar parámetros
    w = w - alpha * dw
    b = b - alpha * db
    
    # Imprimir la pérdida cada 100 iteraciones (para ver el progreso)
    # if i % 10 == 0:
    # print(f"Iteración {i}, Pérdida: {loss}")
```

```python
def predict(X, w, b, threshold=0.5):
    """
    X: datos de entrada (nuevas muestras)
    w: parámetros del modelo (pesos)
    b: sesgo (bias)
    threshold: umbral para tomar decisiones (por defecto 0.5)
    
    Retorna las predicciones (0 o 1)
    """
    z = X @ w + b  # Entrada lineal
    p = sigmoid(z)  # Aplicar sigmoide para obtener probabilidades
    
    # Decidir la clase (0 o 1) usando el umbral
    predictions = (p >= threshold).astype(int)  # 1 si p >= threshold, 0 si no
    return predictions, p
```

```python
X_test = np.array([[0,0],[1,1]])

predict(X_test, w, b)
```
